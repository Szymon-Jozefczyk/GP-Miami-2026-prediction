```python
event_names = []

for year in [2022, 2023, 2024, 2025]:
    schedule = fastf1.get_event_schedule(year, include_testing=False)
    for name in schedule['EventName']:
        event_names.append((year, name))

# Lista unikalnych wartości z informacją o roku
event_names_df = pd.DataFrame(event_names, columns=['year', 'EventName'])
print(event_names_df.to_string())

# Samo unikalne nazwy bez roku (jeśli potrzebujesz płaskiej listy)
unique_names = sorted(set(name for _, name in event_names))
print(f"\nUnikalnych nazw łącznie: {len(unique_names)}")
print(unique_names)
```
