```python
gp_25 = fastf1.get_event_schedule(2025)['EventName']
```

```python
gp_25_list = gp_25.to_list()

print(gp_25_list)

gp_26_list = ['Pre-Season Testing', 'Australian Grand Prix', 'Chinese Grand Prix', 'Japanese Grand Prix']
```

```python
event = fastf1.get_event(2026, 'Australia')
print(event)
```

```python
data_25 = []
form_of_race = ['Sprint Qualifying', 'Sprint', 'Qualifying', 'Race']

for gp in gp_25_list:
    for race in form_of_race:
        try:
            session = fastf1.get_session(2025, gp, race)
            session.load()

            results = session.results.copy()
            results['year'] = 2025

            data_25.append(results)
        except:
            continue


print(data_25)
```

```python
df_25 = pd.concat(data_25, ignore_index=True)
```
